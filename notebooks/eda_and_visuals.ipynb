```python
import os
import pandas as pd
import matplotlib.pyplot as plt

os.chdir(r"C:\Users\Admin\nasa-neo-asteroid-analytics")

df = pd.read_csv("data/processed/neo_clean.csv")

monthly = df.groupby("approach_month").size()
monthly.plot(kind="line", marker="o", title="NEO Close Approaches by Month")
plt.ylabel("Number of approaches")

# add a data label above each point showing its exact value
for x, y in enumerate(monthly.values):
    plt.annotate(str(y), (x, y), textcoords="offset points", xytext=(0, 8), ha="center")

plt.savefig("visuals/monthly_trend.png", bbox_inches="tight")
plt.show()
```

```python
df["avg_diameter_km"].plot(kind="hist", bins=30, title="Distribution of Asteroid Diameters")
plt.xlabel("Average diameter (km)")
plt.savefig("visuals/size_distribution.png", bbox_inches="tight")
plt.show()
```

```python
import seaborn as sns

sns.scatterplot(data=df, x="relative_velocity_kph", y="miss_distance_lunar", hue="is_potentially_hazardous")
plt.title("Velocity vs. Miss Distance")
plt.savefig("visuals/velocity_vs_distance.png", bbox_inches="tight")
plt.show()
```
